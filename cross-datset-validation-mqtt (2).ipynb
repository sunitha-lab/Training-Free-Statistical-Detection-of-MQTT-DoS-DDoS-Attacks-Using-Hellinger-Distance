```python
import numpy as np
from scipy.spatial.distance import euclidean

# ==================================================
# PROBABILITY DISTRIBUTION
# ==================================================

def get_distribution(df):

    counts = df['mqtt.msgtype'].value_counts()

    probs = counts / counts.sum()

    return probs


# ==================================================
# HELLINGER DISTANCE
# ==================================================

def hellinger(p, q):

    keys = sorted(
        list(set(p.index).union(set(q.index)))
    )

    p_vec = np.array(
        [p.get(k, 0) for k in keys]
    )

    q_vec = np.array(
        [q.get(k, 0) for k in keys]
    )

    return euclidean(
        np.sqrt(p_vec),
        np.sqrt(q_vec)
    ) / np.sqrt(2)
```

```python
normal_profile = get_distribution(normal)

results = []

for name, df in attacks.items():

    attack_profile = get_distribution(df)

    hd = hellinger(
        normal_profile,
        attack_profile
    )

    results.append(
        [name, hd]
    )

    print(
        f"{name} --> HD = {hd:.6f}"
    )
```

```python
import numpy as np

def get_numeric_distribution(series, bins=10):

    series = series.dropna()

    # Empty feature
    if len(series) == 0:

        return np.ones(bins) / bins

    hist, _ = np.histogram(

        series,

        bins=bins
    )

    hist = hist.astype(float)

    total = hist.sum()

    if total == 0:

        return np.ones(bins) / bins

    return hist / total
```

```python
from scipy.spatial.distance import euclidean

def hellinger_hist(p, q):

    return euclidean(

        np.sqrt(p),

        np.sqrt(q)

    ) / np.sqrt(2)
```

```python
attacks = {

    "Flood": flood,
    "Bruteforce": bruteforce,
    "Malaria": malaria,
    "Malformed": malformed,
    "Slowite": slowite
}

results_B = []

for name, df in attacks.items():

    # =====================================
    # MESSAGE TYPE HD
    # =====================================

    attack_msgtype = get_distribution(df)

    hd_msgtype = hellinger(

        normal_msgtype,

        attack_msgtype
    )

    # =====================================
    # TOPIC LENGTH HD
    # =====================================

    attack_topic = get_numeric_distribution(

        df['mqtt.topic_len']
    )

    hd_topic = hellinger_hist(

        normal_topic,

        attack_topic
    )

    # =====================================
    # MESSAGE LENGTH HD
    # =====================================

    attack_msglen = get_numeric_distribution(

        df['mqtt.len']
    )

    hd_msglen = hellinger_hist(

        normal_msglen,

        attack_msglen
    )

    # =====================================
    # COMBINED HD
    # =====================================

    hd_combined = (

        hd_msgtype +

        hd_topic +

        hd_msglen

    ) / 3

    results_B.append([

        name,

        hd_msgtype,

        hd_topic,

        hd_msglen,

        hd_combined
    ])

    print(
        f"{name}: "
        f"Combined HD = {hd_combined:.4f}"
    )
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    confusion_matrix,
    f1_score,
    roc_auc_score
)

import numpy as np
import pandas as pd

# ============================================================
# SETTINGS
# ============================================================

WINDOW_SIZE = 100      # packets per statistical window
RANDOM_STATE = 42

# Use all MQTT message types observed in legitimate traffic
# and attack traffic so that every window has the same vector.
all_msgtypes = sorted(
    set(normal['mqtt.msgtype'].dropna().unique()).union(
        *[
            set(df['mqtt.msgtype'].dropna().unique())
            for df in attacks.values()
        ]
    )
)

print("Number of MQTT message types:", len(all_msgtypes))
print(all_msgtypes)


# ============================================================
# CREATE WINDOW-LEVEL MESSAGE-TYPE DISTRIBUTIONS
# ============================================================

def create_windows(df, window_size=WINDOW_SIZE):

    data = df[['mqtt.msgtype']].dropna().reset_index(drop=True)

    distributions = []

    for start in range(0, len(data), window_size):

        window = data.iloc[start:start + window_size]

        # Ignore very small final windows
        if len(window) < window_size:
            continue

        counts = (
            window['mqtt.msgtype']
            .value_counts()
            .reindex(all_msgtypes, fill_value=0)
            .values
            .astype(float)
        )

        probs = counts / counts.sum()

        distributions.append(probs)

    return np.array(distributions)


normal_windows = create_windows(normal)

print("Normal windows:", normal_windows.shape)

for name, df in attacks.items():
    print(name, create_windows(df).shape)
```

```python
# ============================================================
# SPLIT NORMAL WINDOWS
# ============================================================

rng = np.random.default_rng(RANDOM_STATE)

indices = rng.permutation(len(normal_windows))

normal_windows = normal_windows[indices]

n = len(normal_windows)

n_reference = int(0.50 * n)
n_calibration = int(0.25 * n)

reference_windows = normal_windows[:n_reference]

calibration_windows = normal_windows[
    n_reference:n_reference + n_calibration
]

test_normal_windows = normal_windows[
    n_reference + n_calibration:
]

print("Reference windows:", len(reference_windows))
print("Calibration windows:", len(calibration_windows))
print("Held-out normal test windows:", len(test_normal_windows))
```

```python
# ============================================================
# MQTTSET-SPECIFIC BENIGN REFERENCE PROFILE
# ============================================================

reference_profile = reference_windows.mean(axis=0)

# Normalize for numerical safety
reference_profile = (
    reference_profile / reference_profile.sum()
)

print("Reference profile:")
print(reference_profile)

print("Sum =", reference_profile.sum())
```

```python
# ============================================================
# HELLINGER DISTANCE
# ============================================================

def hellinger_vector(p, q):

    return np.sqrt(
        np.sum(
            (np.sqrt(p) - np.sqrt(q)) ** 2
        )
    ) / np.sqrt(2)


def calculate_distances(windows, reference):

    return np.array([
        hellinger_vector(window, reference)
        for window in windows
    ])


calibration_distances = calculate_distances(
    calibration_windows,
    reference_profile
)

print("Calibration HD mean:",
      calibration_distances.mean())

print("Calibration HD std:",
      calibration_distances.std())
```

```python
# ============================================================
# THRESHOLD USING BENIGN CALIBRATION DATA ONLY
# ============================================================

mu = calibration_distances.mean()

sigma = calibration_distances.std()

threshold = mu + 3 * sigma

print("Mean benign calibration HD =", mu)
print("Std benign calibration HD  =", sigma)
print("Detection threshold         =", threshold)
```

```python
# ============================================================
# HELD-OUT NORMAL EVALUATION
# ============================================================

normal_test_distances = calculate_distances(
    test_normal_windows,
    reference_profile
)

normal_predictions = (
    normal_test_distances > threshold
).astype(int)

false_positive_rate = normal_predictions.mean()

print("Held-out normal windows:",
      len(test_normal_windows))

print("False positives:",
      normal_predictions.sum())

print("FPR:",
      false_positive_rate)

print("FPR (%):",
      false_positive_rate * 100)
```

```python
# ============================================================
# ATTACK-WISE EVALUATION
# ============================================================

results = []

for attack_name, attack_df in attacks.items():

    attack_windows = create_windows(attack_df)

    if len(attack_windows) == 0:

        print(
            f"{attack_name}: insufficient packets "
            f"for WINDOW_SIZE={WINDOW_SIZE}"
        )

        continue

    attack_distances = calculate_distances(
        attack_windows,
        reference_profile
    )

    # --------------------------------------------------------
    # Build evaluation set:
    # held-out benign windows + current attack windows
    # --------------------------------------------------------

    y_true = np.concatenate([
        np.zeros(len(normal_test_distances)),
        np.ones(len(attack_distances))
    ])

    distances = np.concatenate([
        normal_test_distances,
        attack_distances
    ])

    y_pred = (
        distances > threshold
    ).astype(int)

    # --------------------------------------------------------
    # METRICS
    # --------------------------------------------------------

    accuracy = accuracy_score(
        y_true,
        y_pred
    )

    precision = precision_score(
        y_true,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        zero_division=0
    )

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1]
    ).ravel()

    fpr = fp / (fp + tn) if (fp + tn) else 0

    auc = roc_auc_score(
        y_true,
        distances
    )

    results.append({
        "Attack": attack_name,
        "Normal Test Windows": len(normal_test_distances),
        "Attack Windows": len(attack_distances),
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1,
        "FPR": fpr,
        "ROC-AUC": auc,
        "TP": tp,
        "TN": tn,
        "FP": fp,
        "FN": fn
    })


results_df = pd.DataFrame(results)

results_df
```

```python
# ============================================================
# FINAL TABLE FOR MANUSCRIPT
# ============================================================

final_table = results_df[
    [
        "Attack",
        "Accuracy",
        "Precision",
        "Recall",
        "FPR",
        "ROC-AUC"
    ]
].copy()

for col in [
    "Accuracy",
    "Precision",
    "Recall",
    "FPR",
    "ROC-AUC"
]:
    final_table[col] = (
        final_table[col] * 100
    ).round(2)


final_table.columns = [
    "Attack",
    "Accuracy (%)",
    "Precision (%)",
    "Recall (%)",
    "FPR (%)",
    "ROC-AUC (%)"
]

print("\nMQTTset Cross-Dataset Validation Results\n")

print(
    final_table.to_string(index=False)
)
```
